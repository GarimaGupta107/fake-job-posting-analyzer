## Fill Missing Values

```python
df.fillna({'location':"Unknown"},inplace=True)
df.fillna({'department':'Not Specified'},inplace=True)
df.fillna({'company_profile':'Not Available'},inplace=True)
df.fillna({'requirements':'Not Specified'},inplace=True)
df.fillna({'benefits':'Not Provided'},inplace=True)
df.fillna({'employment_type':'Not Specified'},inplace=True)
df.fillna({'required_experience':'Not Specified'},inplace=True)
df.fillna({'required_education':'Not Specified'},inplace=True)
df.fillna({'industry':'Not Specified'},inplace=True)
df.fillna({'function':'Not Specified'},inplace=True)
df.fillna({'description':'Not Provided'},inplace=True)
```

## Check For Missing Values

## Formatting

```python
df['title']=df['title'].str.strip()
df['location']=df['location'].str.strip()
df['department']=df['department'].str.strip()
df['company_profile']=df['company_profile'].str.strip()
df['description']=df['description'].str.strip()
df['salary_range']=df['salary_range'].str.strip()
df['requirements']=df['requirements'].str.strip()
df['benefits']=df['benefits'].str.strip()
df['employment_type']=df['employment_type'].str.strip()
df['required_experience']=df['required_experience'].str.strip()
df['required_education']=df['required_education'].str.strip()
df['industry']=df['industry'].str.strip()
df['function']=df['function'].str.strip()


```
